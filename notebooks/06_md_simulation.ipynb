## 1. Setup Simulation Systems

For each of the top 2 docked compounds, prepare the system.

**Note:** This requires GROMACS installed and protein-ligand complex PDB files prepared from the docking results.

````python
# Compound directories
compounds = {
    "compound_1": "../results/md_simulation/compound_1",
    "compound_2": "../results/md_simulation/compound_2",
}

# Write MDP files for each compound
for name, outdir in compounds.items():
    mdp_files = md_sim.write_mdp_files(outdir)
    print(f"{name}: MDP files written to {outdir}")
    for stage, path in mdp_files.items():
        print(f"  {stage}: {path}")
````
